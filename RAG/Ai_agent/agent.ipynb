```python
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
import requests
```

```python
from langchain_community.tools import DuckDuckGoSearchRun

search_tool=DuckDuckGoSearchRun()

result=search_tool.invoke('topnews in india today')
```

```python
llm=ChatOpenAI()
```

```python
llm.invoke("hi")
```

```python
from langchain.agents import craeet_react_agent,AgentExecutor
from langchain import hub
```

```python
# Step ===> Pull the react prompt from the langchain hub
```

```python
prompt=hub.pull("hwchase17/react") # pulls the standard readt agent prompt
```

```python
# step ==> Create the react agent manually with the pulled prompt
```

```python
agent=craeet_react_agent(
    llm=llm,
    tools=[search_tool],
    prompt=prompt
)
```

```python
# step==> wrap the agent executor
```

```python
agent_executor=AgentExecutor(
    agent=agent,
    tools=[search_tool],
    verbose=True
)
# verbose -> whatagent is doing in back will be shown to you 
```

```python
# step===> Invoke 
```

```python
response=agent_executor.invoke({"input":"3 ways to reach goa fom delhi"})
print(response)
```
